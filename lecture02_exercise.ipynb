```python
import pandas as pd
import numpy as np

# World Happiness Report 2023 — representative data
# Source: https://www.kaggle.com/datasets/ajaypalsinghlo/world-happiness-report-2023

df = pd.read_csv('/content/world_happiness_2023.csv')
df.columns = ['Country','Region','Happiness_Score','GDP','Social_Support',
              'Life_Expectancy','Freedom','Generosity','Corruption']


print(f"Dataset: {len(df)} countries, {len(df.columns)} columns")
print(df.head())
```

```python
import plotly.express as px
import plotly.graph_objects as go

# Explore the dataset before you start
print("Regions in dataset:")
print(df['Region'].value_counts())
print("\nScore range:", df['Happiness_Score'].min(), "–", df['Happiness_Score'].max())
print("\nBottom 10 countries:")
print(df.nsmallest(10, 'Happiness_Score')[['Country','Region','Happiness_Score']])
```

```python
# Task 1: Regional comparison bar chart
# -------------------------------------

region_avg = (df.groupby('Region')['Happiness_Score']
              .mean()
              .reset_index()
              .sort_values('Happiness_Score'))

print(region_avg)
```

```python
import plotly.express as px

fig = px.bar(
    region_avg,
    x='Happiness_Score',
    y='Region',
    orientation='h',
    text='Happiness_Score',
    title='Western Europe leads regional happiness, highlighting a clear wellbeing gap',
    labels={
        'Happiness_Score': 'Average Happiness Score',
        'Region': ''
    }
)

fig.update_traces(
    marker_color='steelblue',
    texttemplate='%{text:.2f}',
    textposition='outside'
)

fig.update_layout(
    xaxis=dict(
        range=[0, region_avg['Happiness_Score'].max() * 1.15],
        title='Average Happiness Score'
    ),
    yaxis=dict(
        categoryorder='total ascending'
    ),
    plot_bgcolor='white',
    paper_bgcolor='white',
    font=dict(size=14),
    title_font=dict(size=20),
    margin=dict(l=180, r=80, t=90, b=60)
)

fig.show()
```

```python
top8 = df.nlargest(8, 'Happiness_Score').copy()
```

```python
top8['Group'] = 'Top 8'
```

```python
bottom8 = df.nsmallest(8, 'Happiness_Score').copy()
```

```python
bottom8['Group'] = 'Bottom 8'
```

```python
combined = pd.concat([bottom8, top8]).sort_values('Happiness_Score')
```

```python
global_avg = df['Happiness_Score'].mean()
```

```python
print(f"Global average: {global_avg:.2f}")
```

```python
fig = px.bar(
    combined,
    x='Happiness_Score',
    y='Country',
    orientation='h',
    color='Group',
    text='Happiness_Score',
    color_discrete_map={
        'Top 8': 'green',
        'Bottom 8': 'red'
    },
    title='The happiest countries score far above the least happy countries, revealing a major wellbeing gap',
    labels={
        'Happiness_Score': 'Happiness Score',
        'Country': '',
        'Group': 'Group'
    }
)

fig.update_traces(
    texttemplate='%{text:.2f}',
    textposition='outside'
)

fig.add_vline(
    x=global_avg,
    line_dash='dash',
    line_width=2,
    annotation_text=f'Global average: {global_avg:.2f}',
    annotation_position='top'
)

fig.update_layout(
    xaxis=dict(
        range=[0, combined['Happiness_Score'].max() * 1.20],
        title='Happiness Score'
    ),
    yaxis=dict(
        categoryorder='total ascending'
    ),
    plot_bgcolor='white',
    paper_bgcolor='white',
    font=dict(size=14),
    title_font=dict(size=20),
    margin=dict(l=160, r=100, t=100, b=60)
)

fig.show()
```
